#### **1. Importing libraries**

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
```

### **Part A**

---

#### **Loading Dataset**

```python
df = load_iris()
```

```python
iris_df = pd.DataFrame(df.data, columns=df.feature_names)
iris_df['target'] = df.target

display(iris_df.head())
```

```python
iris_df = iris_df[iris_df['target'].isin([0, 1])]
iris_df['target'].value_counts()
```

```python
print("shape of X", iris_df.shape)
```

```python
iris_df_X = iris_df[['sepal length (cm)','sepal width (cm)']]
iris_df_Y = iris_df['target']
```

```python
display(iris_df_X.head())
```

```python
display(iris_df_Y.head())
```

```python
iris_df_Y = np.where(iris_df_Y == 0, 1, -1)
```

#### **Train/Test Split**

```python
X_train, X_test, Y_train, Y_test = train_test_split(iris_df_X,iris_df_Y,train_size=0.8,random_state=42)
print("shape of X-train", X_train.shape)
print("shape of Y-train", Y_train.shape)
print("shape of X-test", X_test.shape)
print("shape of Y-test", Y_test.shape)
```

#### **Single Layer Perceptron**

```python
perceptron_model = Perceptron(eta0=0.1, max_iter=100, random_state=0)
```

```python
perceptron_model.fit(X_train, Y_train)
Y_pred = perceptron_model.predict(X_test)
```

```python
X_train_pred = perceptron_model.predict(X_train)
```

```python
training_accuracy = accuracy_score(Y_train, X_train_pred)
print(f"Training Accuracy: {training_accuracy:.2f}")
testing_accuracy = accuracy_score(Y_test, Y_pred)
print(f"Testing Accuracy: {testing_accuracy:.2f}")
```

```python
print(f"Coefficients (Weights): {perceptron_model.coef_}")
print(f"Intercept (Bias): {perceptron_model.intercept_}")
```

```python
class_report = classification_report(Y_test, Y_pred)
print("Classification Report:\n", class_report)
```
